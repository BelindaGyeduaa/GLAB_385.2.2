### Python Dictionaries
- Data collection. Keyed collections
- Key: Value pairs
- Any data types

```python
# Defining dictionaries

student_info = {
    "name": "Belinda",
    "age": 30,
    "courses": [],
    "is_enrolled": True
}



# Lists - list of nouns. (list of animals, list of names, list of groceries)
# Dict - single noun and all of the properties describe the noun

# cars = ['honda', 'toyota', 'ford', 'lexus']

# car = {
#     "make": "honda",
#     "model": "civic",
#     "engine": {
#         "cyl": 4,
#         "hp": 100,
#         'size': [1.8, 2, 4]

#     }
# }

print(type(student_info))
```

```python
# Accessing Data from a Dictionary

print(student_info["name"])
print(student_info.get("name"))
```

```python
#Modifying INformation from a dict

# update an existing value
student_info['age'] = student_info['age'] + 1

# Creating new key:value pairs
student_info['grade'] = "A"

# If you call upon a specific property, you can use the methods of that properties data type
student_info['courses'].append('Data Science')
student_info['courses'].extend(['BI', 'Data Engineering'])

# You have access to the methods of the individual data types for each key value pair
student_info['name'] = student_info['name'].upper()

print(student_info)
```

```python
# Removing Properties from a dictionary
del student_info['grade']

# Pop to delete from a dictionary, allows you to save the deleted value. Gets returned by the method
removed_value = student_info.pop('is_enrolled')

print(f'Removed Value: {removed_value}')

print(student_info)
```

```python
# Common Dictionary Methods
print(student_info.keys())  # returned list of keys
print(student_info.values())  # returned list of values
```

```python
# Super useful for looping through an object
print(student_info.items())  # returns list of tuples of the key value pairs


# for pair in student_info.items():
#     print(pair)


# Destructure tuples in a loop
for key, value in student_info.items():
    print(f'{key}: {value}')
```
